## Upload a videoUse the next cell to upload a video file. The uploaded path is stored in `VIDEO_PATH`.

```python
from google.colab import filesfrom pathlib import Pathuploaded = files.upload()if not uploaded:    raise RuntimeError('Upload a video file to continue.')filename = next(iter(uploaded))VIDEO_PATH = str(Path(filename).resolve())print('Using video:', VIDEO_PATH)
```

## Configure LLaVA inference
These settings are **LLaVA-only**. Set `LORA_ADAPTER_PATH` to a LoRA adapter, or keep it `None` to use merged weights.

```python
BASE_MODEL_NAME = 'llavavideo_qwen_7b'MODEL_PATH = 'pritamqu/LLaVA-Video-7B-Qwen2-RRPO-32f'LORA_ADAPTER_PATH = None  # Set to a LoRA adapter repo/path, or keep None to use merged weightsQUESTION = 'Describe this video.'MODEL_MAX_LENGTH = 1024
```

## Run inference

```python
import subprocesscmd = [    'python', 'inference.py',    '--base_model_name', BASE_MODEL_NAME,    '--model-path', MODEL_PATH,    '--video_path', VIDEO_PATH,    '--question', QUESTION,    '--model_max_length', str(MODEL_MAX_LENGTH),]if LORA_ADAPTER_PATH:    cmd.extend(['--model-path2', LORA_ADAPTER_PATH])print('Running:', ' '.join(cmd))subprocess.run(cmd, check=True)
```
