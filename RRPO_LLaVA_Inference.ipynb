```python
# LLaVA-only inference configuration\nBASE_MODEL_NAME = "llavavideo_qwen_7b"\nMODEL_PATH = "pritamqu/LLaVA-Video-7B-Qwen2-RRPO-32f"  # merged RRPO model\nLORA_PATH = None  # e.g. "pritamqu/LLaVA-Video-7B-Qwen2-RRPO-32f-LORA"\nVIDEO_PATH = "sample_video.mp4"\nQUESTION = "Describe this video."\nMODEL_MAX_LENGTH = 1024
```

```python
import subprocess\n\ncmd = [\n    "python", "inference.py",\n    "--base_model_name", BASE_MODEL_NAME,\n    "--model-path", MODEL_PATH,\n    "--video_path", VIDEO_PATH,\n    "--question", QUESTION,\n    "--model_max_length", str(MODEL_MAX_LENGTH),\n]\n\nif LORA_PATH:\n    cmd.extend(["--model-path2", LORA_PATH])\n\nsubprocess.run(cmd, check=True)
```
